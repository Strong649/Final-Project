## Game Logs

```python
import os
import pandas as pd

# Define the directory containing the scraped gamelog CSV files
gamelog_dir = "scraped_gamelogs"  # Change this if needed

# Initialize an empty list to store individual DataFrames
dataframes = []

# Loop through all year folders inside `scraped_gamelogs`
for year in os.listdir(gamelog_dir):
    year_path = os.path.join(gamelog_dir, year)

    # Ensure it's a directory
    if not os.path.isdir(year_path):
        continue

    # Loop through "reg" and "post" subfolders inside each year folder
    for season_type in ["reg", "post"]:
        season_path = os.path.join(year_path, season_type)

        if not os.path.exists(season_path):
            continue  # Skip if the subfolder doesn't exist

        # Load CSV files inside each season-type folder
        for filename in os.listdir(season_path):
            if filename.endswith(".csv"):
                file_path = os.path.join(season_path, filename)

                # Read the CSV file into a pandas DataFrame
                df = pd.read_csv(file_path)

                # Extract metadata from filename
                parts = filename.split("_")
                if len(parts) >= 3:
                    team = parts[1]  # Extract team code
                    season_year = parts[2]  # Extract year
                    df["Team"] = team
                    df["Season"] = season_year
                    df["SeasonType"] = season_type  # "reg" or "post"

                # Append the DataFrame to the list
                dataframes.append(df)

# Combine all DataFrames into one unified dataset
compiled_gamelogs = pd.concat(dataframes, ignore_index=True)

# Display the first few rows
print(compiled_gamelogs.head())

# Optionally, save the merged DataFrame for future use
compiled_gamelogs.to_csv("compiled_gamelogs.csv", index=False)
```

```python
# Drop rows where more than 50% of the columns are NaN
compiled_gamelogs = compiled_gamelogs[compiled_gamelogs.isnull().mean(axis=1) < 0.5]

# Optional: reset index
compiled_gamelogs.reset_index(drop=True, inplace=True)

# Check how many rows were dropped
print(f"✅ Remaining rows after drop: {compiled_gamelogs.shape[0]}")
```

```python
# Create a normalized matchup key (sorted tuple of team and opponent)
compiled_gamelogs['MatchKey'] = compiled_gamelogs.apply(
    lambda row: tuple(sorted([row['Team'], row['Opp']])) + (row['Date'],), axis=1
)

# Sort by that composite key
compiled_gamelogs = compiled_gamelogs.sort_values(by='MatchKey').reset_index(drop=True)

# Drop the helper column if you don't need it
compiled_gamelogs.drop(columns='MatchKey', inplace=True)

# Optional: preview grouped games
print(compiled_gamelogs[['Team', 'Opp', 'Date']].head(10))
```

```python
def create_game_id(df):
    df = df.copy()
    df['GameID'] = df.apply(lambda row: "_".join(sorted([row['Team'], row['Opp']])) + "_" + row['Date'], axis=1)
    return df
```

```python
def create_matchup_table(df):
    df = create_game_id(df)

    # Team perspective
    team_df = df.copy()
    team_df.columns = [f"Team_{col}" if col not in ['GameID'] else col for col in team_df.columns]

    # Opponent perspective
    opp_df = df.copy()
    opp_df.columns = [f"Opp_{col}" if col not in ['GameID'] else col for col in opp_df.columns]

    # Merge both perspectives on GameID
    matchup_df = pd.merge(team_df, opp_df, on='GameID')

    # Drop duplicate matchups (keep one per pair)
    matchup_df = matchup_df[matchup_df['Team_Team'] != matchup_df['Opp_Team']]
    matchup_df = matchup_df.drop_duplicates(subset='GameID').reset_index(drop=True)

    return matchup_df
```

```python
import os
import pandas as pd

# Define the directory containing the scraped gamelog CSV files
advanced_gamelog_dir = "scraped_gamelogs_advanced"  # Change this if needed

# Initialize an empty list to store individual DataFrames
dataframes = []

# Loop through all year folders inside `scraped_gamelogs`
for year in os.listdir(advanced_gamelog_dir):
    year_path = os.path.join(advanced_gamelog_dir, year)

    # Ensure it's a directory
    if not os.path.isdir(year_path):
        continue

    # Loop through "reg" and "post" subfolders inside each year folder
    for season_type in ["reg", "post"]:
        season_path = os.path.join(year_path, season_type)

        if not os.path.exists(season_path):
            continue  # Skip if the subfolder doesn't exist

        # Load CSV files inside each season-type folder
        for filename in os.listdir(season_path):
            if filename.endswith(".csv"):
                file_path = os.path.join(season_path, filename)

                # Read the CSV file into a pandas DataFrame
                df = pd.read_csv(file_path)

                # Extract metadata from filename
                parts = filename.split("_")
                if len(parts) >= 3:
                    team = parts[2]  # Extract team code
                    season_year = parts[3]  # Extract year
                    df["Team"] = team
                    df["Season"] = season_year
                    df["SeasonType"] = season_type  # "reg" or "post"

                # Append the DataFrame to the list
                dataframes.append(df)

# Combine all DataFrames into one unified dataset
compiled_advanced_gamelogs = pd.concat(dataframes, ignore_index=True)

# Display the first few rows
print(compiled_advanced_gamelogs.head())

# Optionally, save the merged DataFrame for future use
compiled_advanced_gamelogs.to_csv("compiled_advanced_gamelogs.csv", index=False)
```
